# Film Maker — idea → animated film (MiniMax H3)
The notebook only sets configuration and runs the pipeline; all logic lives in the `film_maker` package.

**Phase 1** writes an editable `plan.json` (+ readable `plan.md`) — no image/video models load.  
**Phase 2** renders stills → H3 clips (voice-chained waves) → final film with H3's own audio.

```python
import os
from film_maker import (FILM_CONFIG, plan_film, produce_film,
                        regenerate_scenes, list_projects, find_plan)

FILM_CONFIG.update({
    "project_name": "lighthouse_vortex_grok",   # everything for this film lives under this folder
    "film_mode": "story",               # or "educational" (idea = a topic to teach)
    "registers": ["scifi", "drama"],   # first = primary; e.g. ["comedy"], ["horror", "thriller"]
    "content_rating": "teen",          # all_ages | teen | mature
    "tone_notes": "",                  # optional free-text tonal guidance
    "target_minutes": 1.0,
    "cast_size": "small",
    "visual_style": "cgi",               # "watercolor" | "graphic_novel" | "photorealistic" | "cgi" | "anime" | free text
    "camera_style": "slow confident dolly moves, deliberate reframes, no handheld",               # "classical" | "handheld" | "epic" | "tableau" | "kinetic" | "floating"
    "dialogue_style": "Twain-craft dialect vernacular",   # Twain-craft dialect; or "spare" | "rapid_wit" | "naturalistic" | "lyrical" | "hardboiled"
    "image_variants": 3,                # stills per shot; vision pass picks the best
    "voice_chaining": True,             # lock each character's H3 voice via audio refs
    "scene_start_image": False,         # if you want the still as reference-only without pinning frame 1
    "use_reference_stills": False,      # no Z-Image at all; identity via prompt locks
    "use_grok": True,                 # route planning LLM calls to Grok (needs XAI_API_KEY)
    "wgp_repo_dir": os.path.expanduser("~/repos/Wan2GP"),
    "wgp_conda_env": "wan2gp",
})
```

```python
IDEA = "A lighthouse keeper discovers the light is a door, and something on the other side has been reading her letters."
#plan_path = find_plan(FILM_CONFIG)  
plan_path = plan_film(IDEA, FILM_CONFIG)
```
